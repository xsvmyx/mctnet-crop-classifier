###  in this code we are replacing the 1 value (means true) by its real crop id for each crop file

````python
from pathlib import Path

#in this folder , each file contains one type of crop and its name is the name of the crop 
#we faced this problem because we changed the export strategy
#here we exported each crop in a separate file instead of exporting all the crops in one file 
#instead of randomly exporting samples in one file
#because not every crop has the same number across these 2 states.
dossier = Path('../data_fix')

for fichier in dossier.iterdir():
    if fichier.is_file(): 
        print(fichier.name)
````

````python
import pandas as pd
from pathlib import Path


dossier_source = Path('../data_fix')

# keywords are matched in the filename to determine the correct code to apply
mapping_codes = {
    'Soybeans': 1,
    'Corn': 5,
    'Rice': 3,
    'Cotton': 2,
    'Grapes': 69,
    'Almonds': 75,
    'Alfalfa': 36,
    'Pistachio': 76,  
    'Others': 176,
    'Grass': 176      
}

def fix_labels_in_folder():
    fichiers = list(dossier_source.glob('*.csv'))
    print(f"🚀 Début de la correction sur {len(fichiers)} fichiers...")

    for f in fichiers:
       
        code_a_appliquer = None
        nom_fichier = f.name
        
        for keyword, code in mapping_codes.items():
            if keyword in nom_fichier:
                code_a_appliquer = code
                break
        
        if code_a_appliquer is None:
            print(f"⚠️  Impossible de déterminer le code pour : {nom_fichier}")
            continue

        try:
            df = pd.read_csv(f)
            
            if 'cropland' in df.columns:
               
                df['cropland'] = code_a_appliquer
                
                
                df.to_csv(f, index=False)
                print(f"✅ Corrigé : {nom_fichier} -> ID {code_a_appliquer}")
            else:
                print(f"❓ Colonne 'cropland' absente dans : {nom_fichier}")
                
        except Exception as e:
            print(f"❌ Erreur sur {nom_fichier} : {e}")

if __name__ == "__main__":
    fix_labels_in_folder()
    print("\n✨ Traitement terminé. Tous tes '1' sont maintenant des ID CDL corrects.")
````
