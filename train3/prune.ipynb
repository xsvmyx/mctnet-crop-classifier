```python
import torch

# Charger le fichier
state_dict = torch.load('./models/model_california_all.pth', map_location=torch.device('cpu'))

# Lister toutes les couches pour repérer celles que tu veux analyser
for layer_name, weights in state_dict.items():
    print(f"Couche: {layer_name} | Taille: {weights.size()}")
```

```python
for layer_name, weights in state_dict.items():
    if 'weight' in layer_name:  # On se concentre sur les poids, pas les bias
        # Calculer le pourcentage de poids "morts" (proches de 0)
        threshold = 1e-5
        zeros = (torch.abs(weights) < threshold).sum().item()
        total = weights.numel()
        percent_zeros = (zeros / total) * 100
        
        print(f"{layer_name} : {percent_zeros:.2f}% de poids presque nuls")
```
